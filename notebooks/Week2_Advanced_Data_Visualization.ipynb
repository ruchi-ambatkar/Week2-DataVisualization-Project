# Load the dataset

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
from pathlib import Path

# Project paths
PROJECT_DIR = Path.cwd().parent
DATA_PATH = PROJECT_DIR / "data" / "2019.csv"
OUTPUT_DIR = PROJECT_DIR / "visualizations"
OUTPUT_DIR.mkdir(exist_ok=True)

# Load dataset
df = pd.read_csv(DATA_PATH)

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
display(df.head())
print("\nColumn names:")
print(df.columns.tolist())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

sns.set_theme(style="whitegrid", palette="viridis")

OUTPUT_DIR = Path.cwd().parent / "visualizations"
OUTPUT_DIR.mkdir(exist_ok=True)

print("Visualization folder:", OUTPUT_DIR)
```

```python
plt.figure(figsize=(10, 6))

sns.histplot(data=df, x="Score", bins=15, kde=True)

plt.title("Distribution of Happiness Scores Across Countries",
          fontsize=15, fontweight="bold")
plt.xlabel("Happiness Score")
plt.ylabel("Number of Countries")
plt.tight_layout()

plt.savefig(OUTPUT_DIR / "01_happiness_distribution.png", dpi=300)
plt.show()
```

This histogram shows how happiness scores are distributed across the 156 countries. The concentration of bars indicates the score range in which most countries fall. The KDE curve helps reveal the overall distribution pattern.

```python
plt.figure(figsize=(10, 6))

sns.regplot(
    data=df,
    x="GDP per capita",
    y="Score",
    scatter_kws={"alpha": 0.7, "s": 55},
    line_kws={"color": "red"}
)

plt.title("Relationship Between GDP per Capita and Happiness",
          fontsize=15, fontweight="bold")
plt.xlabel("GDP per Capita")
plt.ylabel("Happiness Score")
plt.tight_layout()

plt.savefig(OUTPUT_DIR / "02_gdp_vs_happiness.png", dpi=300)
plt.show()
```

This scatter plot investigates the relationship between economic conditions and reported happiness.Observe whether the trend line slopes upward or downward and whether countries cluster around it. An upward trend suggests that higher GDP per capita is associated with higher happiness scores, but it does not prove causation.

```python
plt.figure(figsize=(12, 7))

top_countries = df.nlargest(15, "Score").sort_values("Score")

sns.barplot(
    data=top_countries,
    x="Score",
    y="Country or region"
)

plt.title("Top 15 Happiest Countries",
          fontsize=15, fontweight="bold")
plt.xlabel("Happiness Score")
plt.ylabel("Country")
plt.tight_layout()

plt.savefig(OUTPUT_DIR / "03_top_happiest_countries.png", dpi=300)
plt.show()
```

This horizontal bar chart compares the 15 countries with the highest happiness scores. Finland, Denmark, and Norway appear among the top five in your dataset. The chart makes the differences between leading countries easier to compare.

```python
numeric_df = df.select_dtypes(include="number")

plt.figure(figsize=(11, 8))

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5,
    vmin=-1,
    vmax=1
)

plt.title("Correlation Between Happiness Factors",
          fontsize=15, fontweight="bold")
plt.tight_layout()

plt.savefig(OUTPUT_DIR / "04_correlation_heatmap.png", dpi=300)
plt.show()
```

The heatmap displays the strength and direction of relationships among numerical variables. Values close to +1 indicate strong positive correlations, values near -1 indicate strong negative correlations, and values near 0 indicate weak linear relationships. Pay particular attention to the relationships with Score.

```python
plt.figure(figsize=(12, 7))

bottom_countries = df.nsmallest(15, "Score").sort_values("Score")

sns.barplot(
    data=bottom_countries,
    x="Score",
    y="Country or region"
)

plt.title("15 Countries with the Lowest Happiness Scores",
          fontsize=15, fontweight="bold")
plt.xlabel("Happiness Score")
plt.ylabel("Country")
plt.tight_layout()

plt.savefig(OUTPUT_DIR / "05_lowest_happiness_scores.png", dpi=300)
plt.show()
```

This chart highlights countries with the lowest reported happiness scores in the dataset. It shows the differences among countries at the lower end of the happiness ranking. Compare this chart with Visualization 3 to understand the spread between the highest and lowest scores.

```python
plt.figure(figsize=(11, 7))

sns.scatterplot(
    data=df,
    x="GDP per capita",
    y="Score",
    size="Social support",
    hue="Healthy life expectancy",
    sizes=(40, 350),
    palette="viridis",
    alpha=0.8,
    legend="brief"
)

plt.title("Happiness Across Economic, Social and Health Factors",
          fontsize=15, fontweight="bold")
plt.xlabel("GDP per Capita")
plt.ylabel("Happiness Score")
plt.tight_layout()

plt.savefig(OUTPUT_DIR / "06_multifactor_analysis.png", dpi=300)
plt.show()
```

This bubble chart combines GDP per capita, happiness score, social support, and healthy life expectancy. Bubble size represents social support, while color represents healthy life expectancy. This allows you to explore several variables simultaneously without assuming that any single factor determines happiness.

```python

from docx import Document
from docx.shared import Inches, Pt
from docx.enum.text import WD_ALIGN_PARAGRAPH
from docx.enum.section import WD_SECTION
from pathlib import Path

# Project paths
PROJECT_DIR = Path.cwd().parent
OUTPUT_DIR = PROJECT_DIR / "visualizations"
REPORT_DIR = PROJECT_DIR / "report"
REPORT_DIR.mkdir(exist_ok=True)

report_path = REPORT_DIR / "Week2_Advanced_Data_Visualization_Report.docx"

# Create document
doc = Document()

# Set default font
styles = doc.styles
styles["Normal"].font.name = "Calibri"
styles["Normal"].font.size = Pt(11)

# Title page
title = doc.add_heading(
    "Global Happiness: Data Visualization and Storytelling",
    0
)
title.alignment = WD_ALIGN_PARAGRAPH.CENTER

p = doc.add_paragraph(
    "Week 2 Data Science Project\n"
    "Advanced Data Visualization and Storytelling with Python"
)
p.alignment = WD_ALIGN_PARAGRAPH.CENTER

doc.add_paragraph()
p = doc.add_paragraph("Dataset: World Happiness Report 2019")
p.alignment = WD_ALIGN_PARAGRAPH.CENTER

p = doc.add_paragraph("Tools: Python, Pandas, Matplotlib and Seaborn")
p.alignment = WD_ALIGN_PARAGRAPH.CENTER

doc.add_page_break()

# Introduction
doc.add_heading("1. Introduction", level=1)
doc.add_paragraph(
    "Data visualization transforms numerical data into visual formats "
    "that help readers identify patterns, compare values, and understand "
    "relationships. This project explores the World Happiness Report 2019 "
    "dataset to investigate happiness scores and their relationships with "
    "economic, social, and health-related factors."
)

# Objectives
doc.add_heading("2. Objectives", level=1)
objectives = [
    "Understand the distribution of happiness scores.",
    "Explore the relationship between GDP per capita and happiness.",
    "Compare countries with high and low happiness scores.",
    "Examine correlations among numerical variables.",
    "Explore multiple happiness-related factors in a single visualization.",
    "Communicate findings through clear charts and captions."
]
for item in objectives:
    doc.add_paragraph(item, style="List Bullet")

# Dataset
doc.add_heading("3. Dataset Description", level=1)
doc.add_paragraph(
    f"The dataset contains {len(df)} records and {len(df.columns)} columns "
    "after duplicate removal. The columns include overall rank, country or "
    "region, happiness score, GDP per capita, social support, healthy life "
    "expectancy, freedom to make life choices, generosity, and perceptions "
    "of corruption."
)
doc.add_paragraph(
    "The dataset was checked for missing values and duplicate records. "
    "The cleaning checks showed no missing values in the displayed columns."
)

# Methodology
doc.add_heading("4. Methodology", level=1)
steps = [
    "Loaded the dataset using Pandas.",
    "Removed extra whitespace from column names.",
    "Removed duplicate records.",
    "Checked missing values and inspected the data types.",
    "Created six visualizations using Matplotlib and Seaborn.",
    "Saved the charts as high-resolution PNG files.",
    "Interpreted each chart to support data-driven storytelling."
]
for item in steps:
    doc.add_paragraph(item, style="List Number")

# Visualization explanations
charts = [
    (
        "5.1 Happiness Score Distribution",
        "01_happiness_distribution.png",
        "The histogram displays the distribution of happiness scores. "
        "The bars show how many countries fall within each score interval, "
        "while the KDE curve illustrates the overall distribution pattern."
    ),
    (
        "5.2 GDP per Capita vs Happiness",
        "02_gdp_vs_happiness.png",
        "The scatter plot compares GDP per capita with happiness scores. "
        "The fitted trend line helps show whether these variables have a "
        "positive or negative linear association. Association alone does "
        "not establish causation."
    ),
    (
        "5.3 Top 15 Happiest Countries",
        "03_top_happiest_countries.png",
        "This horizontal bar chart compares the 15 countries with the "
        "highest happiness scores, making differences between their scores "
        "easy to observe."
    ),
    (
        "5.4 Correlation Heatmap",
        "04_correlation_heatmap.png",
        "The heatmap displays pairwise correlations among numerical "
        "variables. Positive values indicate variables that tend to "
        "increase together, while negative values indicate an inverse "
        "linear relationship."
    ),
    (
        "5.5 Countries with the Lowest Happiness Scores",
        "05_lowest_happiness_scores.png",
        "This chart compares the 15 countries with the lowest happiness "
        "scores and highlights variation among countries at the lower end "
        "of the distribution."
    ),
    (
        "5.6 Multiple Happiness Factors",
        "06_multifactor_analysis.png",
        "The bubble chart combines GDP per capita and happiness score, "
        "with bubble size representing social support and color representing "
        "healthy life expectancy. It helps explore several variables at once."
    )
]

doc.add_heading("5. Visualizations and Interpretation", level=1)

for heading, filename, explanation in charts:
    doc.add_heading(heading, level=2)
    image_path = OUTPUT_DIR / filename

    if image_path.exists():
        doc.add_picture(str(image_path), width=Inches(6.0))
        doc.paragraphs[-1].alignment = WD_ALIGN_PARAGRAPH.CENTER
    else:
        doc.add_paragraph(f"Chart file not found: {filename}")

    p = doc.add_paragraph()
    p.add_run("Caption: ").bold = True
    p.add_run(explanation)

# Overall findings
doc.add_heading("6. Key Findings", level=1)

highest = df.nlargest(5, "Score")[["Country or region", "Score"]]
lowest = df.nsmallest(5, "Score")[["Country or region", "Score"]]
gdp_corr = df["GDP per capita"].corr(df["Score"])
social_corr = df["Social support"].corr(df["Score"])
health_corr = df["Healthy life expectancy"].corr(df["Score"])

doc.add_paragraph(
    f"The five highest-scoring countries in this dataset include "
    f"{', '.join(highest['Country or region'].astype(str).tolist())}. "
    "The lowest-scoring countries can be reviewed in the corresponding "
    "bar chart."
)
doc.add_paragraph(
    f"The Pearson correlation between GDP per capita and happiness score "
    f"is {gdp_corr:.2f}. The correlation between social support and "
    f"happiness score is {social_corr:.2f}, while the correlation between "
    f"healthy life expectancy and happiness score is {health_corr:.2f}. "
    "These statistics describe linear associations in this dataset and "
    "should not be interpreted as proof of cause and effect."
)

# Implications
doc.add_heading("7. Implications", level=1)
doc.add_paragraph(
    "The visualizations provide a structured way to compare happiness "
    "scores and investigate their associations with economic, social, "
    "and health-related indicators. Policymakers and researchers could use "
    "such comparisons to identify areas for further investigation. However, "
    "the charts alone cannot establish why a country has a particular score "
    "or prove that changing one indicator will directly change happiness."
)

# Limitations
doc.add_heading("8. Limitations", level=1)
doc.add_paragraph(
    "The analysis describes one dataset from 2019 and does not establish "
    "changes over time. Correlations may be affected by other factors, and "
    "the available indicators may not capture every aspect of well-being. "
    "The results should therefore be interpreted as exploratory findings."
)

# Conclusion
doc.add_heading("9. Conclusion", level=1)
doc.add_paragraph(
    "This project demonstrates how Python visualization libraries can "
    "turn a structured dataset into meaningful charts and explanations. "
    "The six visualizations examine score distributions, country rankings, "
    "relationships, correlations, and multiple factors together. Combining "
    "visual evidence with cautious interpretation makes the analysis easier "
    "to understand and communicate."
)

# Save report
doc.save(report_path)

print("Report created successfully!")
print("Saved to:", report_path)
```
